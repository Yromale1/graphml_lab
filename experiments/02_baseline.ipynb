```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
from graphlab.data import load_and_split_features
from graphlab.models import build_lr_baseline, build_rf_baseline
from graphlab.evaluation import evaluate_binary_classifier
```

```python
X_train, y_train, X_val, y_val, X_test, y_test = load_and_split_features("../data/txs_features.csv", "../data/txs_classes.csv")
print(len(X_train), len(y_train))
print(len(X_val), len(y_val))
print(len(X_test), len(y_test))
```

```python
lr_model = build_lr_baseline()
```

```python
lr_model.fit(X_train, y_train)
```

```python
rf_model = build_rf_baseline()
```

```python
rf_model.fit(X_train, y_train)
```

```python
val_result_lr = evaluate_binary_classifier(lr_model, X_val, y_val)
val_result_rf = evaluate_binary_classifier(rf_model, X_val, y_val)
```

```python
print("Validation Logistic Regression")
print(val_result_lr["classification_report"])
print("Validation Random Forest Classifier")
print(val_result_lr["classification_report"])
```

```python
results_table = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "PR-AUC": [
        val_result_lr["pr_auc"],
        val_result_rf["pr_auc"]
    ],
    "ROC-AUC": [
        val_result_lr["roc_auc"],
        val_result_rf["roc_auc"]
    ]
})

print(results_table.to_string(index=False))
```

```python
cm_lr = val_result_lr["confusion_matrix"]
cm_rf = val_result_rf["confusion_matrix"]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    cmap="coolwarm",
    xticklabels=["Illicit", "Licit"],
    yticklabels=["Illicit", "Licit"],
    cbar=False,
    ax=axes[0],
)

axes[0].set_xlabel("Predicted label")
axes[0].set_ylabel("True label")
axes[0].set_title("Logistic Regression")

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="coolwarm",
    xticklabels=["Illicit", "Licit"],
    yticklabels=["Illicit", "Licit"],
    cbar=False,
    ax=axes[1],
)

axes[1].set_xlabel("Predicted label")
axes[1].set_ylabel("True label")
axes[1].set_title("Random Forest")

plt.suptitle("Confusion Matrices — Validation Set", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
test_result_lr = evaluate_binary_classifier(lr_model, X_test, y_test)
test_result_rf = evaluate_binary_classifier(rf_model, X_test, y_test)
```

```python
print("Test Logistic Regression")
print(test_result_lr["classification_report"])
print("Test Random Forest Classifier")
print(test_result_lr["classification_report"])
```

```python
results_table = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest"],
    "PR-AUC": [
        test_result_lr["pr_auc"],
        test_result_rf["pr_auc"]
    ],
    "ROC-AUC": [
        test_result_lr["roc_auc"],
        test_result_rf["roc_auc"]
    ]
})

print(results_table.to_string(index=False))
```

```python
cm_lr = test_result_lr["confusion_matrix"]
cm_rf = test_result_rf["confusion_matrix"]

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.heatmap(
    cm_lr,
    annot=True,
    fmt="d",
    cmap="coolwarm",
    xticklabels=["Illicit", "Licit"],
    yticklabels=["Illicit", "Licit"],
    cbar=False,
    ax=axes[0],
)

axes[0].set_xlabel("Predicted label")
axes[0].set_ylabel("True label")
axes[0].set_title("Logistic Regression")

sns.heatmap(
    cm_rf,
    annot=True,
    fmt="d",
    cmap="coolwarm",
    xticklabels=["Illicit", "Licit"],
    yticklabels=["Illicit", "Licit"],
    cbar=False,
    ax=axes[1],
)

axes[1].set_xlabel("Predicted label")
axes[1].set_ylabel("True label")
axes[1].set_title("Random Forest")

plt.suptitle("Confusion Matrices — Test Set", fontsize=14)
plt.tight_layout()
plt.show()
```
